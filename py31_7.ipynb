```python
sub1 = int(input("Enter marks of Maths: "))
sub2 = int(input("Enter marks of English: "))
sub3 = int(input("Enter marks of Science: "))

total = sub1 + sub2 + sub3
average = total / 3

print("Total Marks =", total)
print("Average Marks =", average)

if average >= 90:
    print("Grade: A")
elif average >= 80:
    print("Grade: B")
elif average >= 70:
    print("Grade: C")
elif average >= 60:
    print("Grade: D")
elif average >= 40:
    print("Grade: E")
else:
    print("Grade: F")
```

```python
sub1 = int(input("Enter marks of C: "))
sub2 = int(input("Enter marks of C++: "))
sub3 = int(input("Enter marks of Java: "))

total = sub1 + sub2 + sub3
average = total / 3

print("Total Marks =", total)
print("Average Marks =", average)

if average >= 91 and average<=100:
    print("Grade: O")
elif average >= 81 and average<=90:
    print("Grade: A++")
elif average >= 71 and average<=80:
    print("Grade: A")
elif average >= 61 and average<=70:
    print("Grade: B+")
elif average >= 51 and average<=60:
    print("Grade: C+")
elif average >= 41 and average<=50:
    print("Grade: C")
elif average >= 31 and average<=40:
    print("Grade: D")
else:
    print("Grade: F")
```
