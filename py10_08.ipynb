```python
print("Welcome to IDBI Bank home page")
print("Select one operation:")
print('''[1]Deposit
[2]Withdraw
[3]Check Balance
[4]Exit''')

bal=2500
acc_pin=1234

def deposit():
  global bal
  p=int(input("Enter your PIN:"))
  if p==acc_pin:
    print("Pin Verified")
    d=float(input("Enter balance to deposited:"))
    if d>=0 :
      print("Balance deposited successfully!")
      bal+=d
  else:
    print("Wrong PIN, try again!")


def withdraw():
  global bal
  p=int(input("Enter your PIN:"))
  if p==acc_pin:
    print("Pin Verified")
    w=float(input("Enter balance to withdraw:"))
    if w<bal :
      print("Withdrawal Successful. Amount:",w)
      print("Please collect your cash.")
      bal-=w
    else:
      print("Withdrawal Amount exceeding current Balance, Please deposit more to withdraw.")
  else:
    print("Wrong PIN, try again!")


def checkBal(bal):
   p=int(input("Enter your PIN:"))
   if p==acc_pin:
    print("Pin Verified")
    print("Current Balance:",bal)

   else:
    print("Wrong PIN, try again!")


ch=int(input("Enter Choice:"))
if ch==1:
  deposit()

elif ch==2:
  withdraw()

elif ch==3:
  checkBal(bal)

elif ch==4:
  print("Thank you for using IDBI Bank")

else:
  print("Choice not Recognized,please enter a valid choice")




```

```python
print("Welcome to PayFellow, your Payroll Management System")
print("Select from the available options")
print('''[1] Gross Salary Calculation
[2] Calculate Deductions
[3] Calculate Net Salary
[4] View Salary Slip Summary
[5] Exit''')


name = "Amit Roy"
gross_salary = 56090
deductions = 2000
net_salary = 54090


def GrossSal():
    global gross_salary
    basic = float(input("Enter Basic Salary: "))
    da = float(input("Enter DA: "))
    hra = float(input("Enter HRA: "))
    total_salary = basic + da + hra
    if total_salary > 80000 and total_salary < 100000:
        tax = total_salary * 0.30
    elif total_salary > 50000:
        tax = total_salary * 0.10
    else:
        tax = 0
    gross_salary = total_salary - tax
    print("Your Gross Salary is:", gross_salary)


def Deduction():
    global deductions
    if gross_salary == 0:
        print("Please calculate Gross Salary first.")
        return
    ch = int(input("Have you opted for Provident Fund? [1] Yes [2] No: "))
    if ch == 1:
        pf = gross_salary * 0.05
        deductions = pf
        print("Your total deduction is:", deductions)
    elif ch == 2:
        deductions = 0
        print("Your total deduction is:", deductions)
    else:
        print("Please enter a valid choice")


def netSal():
    global net_salary
    net_salary = gross_salary - deductions
    print("Your Net Salary is:", net_salary)


def salarySlip():
    print("\n--- SALARY SLIP ---")
    print("Employee Name:", name)
    print("Gross Salary :", gross_salary)
    print("Deductions   :", deductions)
    print("Net Salary   :", net_salary)
    print("-------------------")


ch = int(input("Enter your choice: "))
if ch == 1:
    GrossSal()
elif ch == 2:
    Deduction()
elif ch == 3:
    netSal()
elif ch == 4:
    salarySlip()
elif ch == 5:
    print("Thank you for using PayFellow!")
else:
    print("Choice not recognized, please enter a valid choice")
```
